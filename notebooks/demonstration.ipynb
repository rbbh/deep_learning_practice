# CIFAR-10 Classification with CNN and Vision Transformer

This notebook demonstrates the classification performance of a simple VGG-like CNN and a simple Vision Transformer (ViT) on the CIFAR-10 dataset.

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

from data.data_loader import Preprocessor
from models.cnn_model import SimpleVGG
from models.vit_model import SimpleViT
from training.trainer import Trainer
from training.tester import Tester
```

## 1. Data Loading and Preprocessing

```python
preprocessor = Preprocessor()
train_loader, test_loader = preprocessor.get_data_loaders()
```

## 2. CNN Model

```python
print("Training CNN Model...")
cnn_model = SimpleVGG()
cnn_trainer = Trainer(cnn_model, train_loader, epochs=5)
cnn_trainer.train()
```

```python
print("Testing CNN Model...")
cnn_tester = Tester(cnn_model, test_loader)
y_true_cnn, y_pred_cnn = cnn_tester.test()
cnn_tester.analyze(y_true_cnn, y_pred_cnn)
```

## 3. Vision Transformer Model

```python
print("Training Vision Transformer Model...")
vit_model = SimpleViT()
vit_trainer = Trainer(vit_model, train_loader, epochs=5)
vit_trainer.train()
```

```python
print("Testing Vision Transformer Model...")
vit_tester = Tester(vit_model, test_loader)
y_true_vit, y_pred_vit = vit_tester.test()
vit_tester.analyze(y_true_vit, y_pred_vit)
```
